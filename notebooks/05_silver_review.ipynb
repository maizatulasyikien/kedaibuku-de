1. Quarantine versus delete. Why keep bad orders in a separate file rather than deleting
them? Who would check that file at a real company?
Quarantining allows you to preserve invalid data for auditing, troubleshooting, and debugging without letting it pollute your clean analytical ("silver") layer. If you deleted them entirely, you would lose visibility into system bugs, fraudulent transactions, or upstream data collection failures. Data Engineers (to fix pipeline bugs and schema issues), Data Quality Analysts, or Customer Operations / Fraud Teams (to investigate why specific customer orders failed or were rejected).

2. Hash versus mask. email_hash lets analysts count unique customers; email_masked is for
display. Why would hashing without a salt be weak for emails? Email addresses have a relatively small and predictable space of common values. If you hash an email without a secure random "salt," attackers or analysts can use rainbow tables (precomputed tables of hashes) or brute-force dictionary attacks to easily reverse or guess the original email addresses from their hashes.

3. Data minimisation. Name, phone and MyKad never reach silver. Which PDPA 2010 principle
does this follow, and why is age_group safer than birth_year? It follows the Retention Principle and the Notice and Choice / Purpose Limitation principles under the Personal Data Protection Act 2010 (Malaysia), which dictate that personal data should not be kept longer than necessary for the fulfillment of the purpose for which it was processed. An exact birth year combined with other demographic or location data can still potentially re-identify an individual (quasi-identifier risk). An age_group (like 25–34) provides higher privacy by generalizing the data while still allowing demographic or marketing analysis.


4. Parquet versus CSV. Load ../data/lake/silver/orders.parquet with
pd.read_parquet() and check .dtypes. What does Parquet remember that CSV would
lose? Parquet preserves the exact data types of each column (such as integers, floats, booleans, and precise timestamp/datetime formats) and structural schemas. CSV files store everything as plain text, meaning pandas has to guess data types on load, often converting missing values or leading zeros incorrectly and losing timezone or type fidelity.
